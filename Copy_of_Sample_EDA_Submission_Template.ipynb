# **Project Summary -**

his project focuses on executing an in-depth Exploratory Data Analysis (EDA) on the Amazon Prime Video dataset containing comprehensive metadata on movies and TV shows available on the streaming platform. The primary objective of this study is to uncover underlying patterns, content distribution strategies, temporal trends, geographical dominance, and consumer preferences to inform content acquisition and audience engagement strategies. The project began with a thorough data preprocessing and cleaning phase, where raw data was systematically transformed into a structured format suitable for analysis. Missing values across key features such as directors, main cast members, and country of origin were systematically evaluated and handled using logical imputation techniques and placeholder values. Furthermore, duplicate records were identified and purged, while date-related attributes were standardized to enable accurate time-series modeling. Once the data was cleaned, univariate, bivariate, and multivariate analysis were conducted using powerful statistical and visualization libraries, specifically Matplotlib and Seaborn. The exploratory analysis revealed several fundamental insights about the Amazon Prime Video catalog. First, a quantitative comparison between content types demonstrated that movies comprise a significantly larger proportion of the platform's overall library compared to TV shows, pointing toward a historically movie-centric acquisition strategy. Second, a temporal analysis of release years showed an exponential increase in total content added to the platform over the last decade, reflecting the global boom in digital streaming adoption and rapid catalog expansion. Third, geographical profiling highlighted the United States and India as the leading global contributors in terms of total content volume, with substantial contributions coming from European and Asian markets as well, showcasing the platform's focus on both global blockbusters and localized regional content. Fourth, evaluating maturity ratings showed a heavy concentration of content categorized under adult and young adult ratings (such as TV-MA, R, and PG-13), indicating a demographic strategy tailored primarily toward mature audiences while maintaining a steady foundation of family-friendly titles. Finally, analyzing genres revealed that Drama, Comedy, Action, and Documentaries dominate the catalog, serving as the core drivers of subscriber retention. Overall, this EDA provides a strategic framework for decision-makers, offering data-driven insights to balance the content mix between TV series and films, identify underserved geographical markets, optimize localized licensing, and guide future content investments to maximize user satisfaction and global platform growth.

## ***4. Data Vizualization, Storytelling & Experimenting with charts : Understand the relationships between variables***

#### Chart - 1

````python
import matplotlib.pyplot as plt
import seaborn as sns

# Chart - 1: Count of Movies vs TV Shows
plt.figure(figsize=(6, 4))
sns.countplot(data=titles_df, x='type', palette='Set2')
plt.title('Distribution of Movies vs TV Shows on Amazon Prime')
plt.xlabel('Type')
plt.ylabel('Count')
plt.show()
````

#### Chart - 2

````python
import matplotlib.pyplot as plt
import seaborn as sns

# Chart - 2: Number of Releases over Years
plt.figure(figsize=(12, 6))
sns.histplot(data=titles_df, x='release_year', bins=30, kde=True, color='purple')
plt.title('Distribution of Content Release Years on Amazon Prime')
plt.xlabel('Release Year')
plt.ylabel('Count')
plt.show()
````

#### Chart - 3

````python
import matplotlib.pyplot as plt
import seaborn as sns

# Chart - 3: Distribution of IMDb Scores
plt.figure(figsize=(10, 5))
sns.histplot(data=titles_df, x='imdb_score', bins=20, kde=True, color='teal')
plt.title('Distribution of IMDb Scores on Amazon Prime')
plt.xlabel('IMDb Score')
plt.ylabel('Count')
plt.show()
````

#### Chart - 4

````python
import matplotlib.pyplot as plt
import seaborn as sns

# Chart - 4: Distribution of Movie Runtimes
plt.figure(figsize=(10, 5))
sns.histplot(data=titles_df[titles_df['type'] == 'MOVIE'], x='runtime', bins=30, kde=True, color='coral')
plt.title('Distribution of Movie Runtimes on Amazon Prime')
plt.xlabel('Runtime (Minutes)')
plt.ylabel('Count')
plt.show()
````

#### Chart - 5

````python
import matplotlib.pyplot as plt
import seaborn as sns

# Chart - 5: Top 10 Genres
plt.figure(figsize=(10, 5))
top_genres = titles_df['genres'].value_counts().head(10)
sns.barplot(x=top_genres.values, y=top_genres.index, palette='magma')
plt.title('Top 10 Most Frequent Genres on Amazon Prime')
plt.xlabel('Count')
plt.ylabel('Genre')
plt.show()
````

#### Chart - 6

````python
import matplotlib.pyplot as plt
import seaborn as sns

# Chart - 6: Distribution of Age Certifications
plt.figure(figsize=(10, 5))
sns.countplot(data=titles_df, y='age_certification', order=titles_df['age_certification'].value_counts().index, palette='crest')
plt.title('Distribution of Content by Age Certification on Amazon Prime')
plt.xlabel('Count')
plt.ylabel('Age Certification')
plt.show()
````

#### Chart - 7

````python
import matplotlib.pyplot as plt
import seaborn as sns

# Chart - 7: Distribution of TMDB Popularity
plt.figure(figsize=(10, 5))
sns.histplot(data=titles_df, x='tmdb_popularity', bins=30, kde=True, color='royalblue')
plt.title('Distribution of TMDB Popularity on Amazon Prime')
plt.xlabel('TMDB Popularity')
plt.ylabel('Count')
plt.show()
````

 I selected a histogram with a KDE plot because it is ideal for analyzing the distribution and spread of right-skewed, continuous numerical metrics like popularity scores.

````python
import matplotlib.pyplot as plt
import seaborn as sns

# Chart - 8: IMDb Score vs TMDB Score Correlation
plt.figure(figsize=(8, 6))
sns.scatterplot(data=titles_df, x='imdb_score', y='tmdb_score', hue='type', alpha=0.7)
plt.title('IMDb Score vs TMDB Score Distribution')
plt.xlabel('IMDb Score')
plt.ylabel('TMDB Score')
plt.show()
````

#### Chart - 10

````python
import matplotlib.pyplot as plt
import seaborn as sns

# Chart - 10: Trend of IMDb Scores over Release Years
plt.figure(figsize=(10, 5))
yearly_scores = titles_df.groupby('release_year')['imdb_score'].mean().reset_index()
sns.lineplot(data=yearly_scores, x='release_year', y='imdb_score', color='darkorange', marker='o')
plt.title('Average IMDb Score Trend Over Release Years')
plt.xlabel('Release Year')
plt.ylabel('Average IMDb Score')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show
````

#### Chart - 14 - Correlation Heatmap

````python
import matplotlib.pyplot as plt
import seaborn as sns

# Chart - 14: Correlation Heatmap
plt.figure(figsize=(8, 6))
numeric_df = titles_df.select_dtypes(include=['float64', 'int64'])
correlation = numeric_df.corr()
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Heatmap of Numerical Variables')
plt.show()
````

#### Chart - 15 - Pair Plot

````python
import matplotlib.pyplot as plt
import seaborn as sns

# Chart - 15: Pair Plot
plt.figure(figsize=(10, 10))
# Select numerical columns and include the 'type' column for hue grouping
numeric_cols = titles_df.select_dtypes(include=['float64', 'int64']).columns.tolist()
if 'type' in titles_df.columns:
    plot_data = titles_df[numeric_cols + ['type']].dropna()
    sns.pairplot(plot_data, hue='type', palette='husl')
else:
    plot_data = titles_df[numeric_cols].dropna()
    sns.pairplot(plot_data, palette='husl')
plt.show()
````
